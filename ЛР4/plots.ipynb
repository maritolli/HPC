```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel('results.xlsx', sheet_name='plot_data (2)')

df.columns = df.columns.astype(str).str.strip().str.lower()

if not {'x', 'y', 'approximation'}.issubset(df.columns):
    print(f"Доступные колонки: {list(df.columns)}")
    raise ValueError("Не найдены нужные колонки: x, y, approximation")

plt.figure(figsize=(12, 7))

plt.plot(df['x'], df['y'], 
         marker='o', markersize=3, linestyle='-', 
         label='Y (фактические)', color='steelblue', linewidth=0.05)

plt.scatter(df['x'], df['approximation'], 
            color='red', s=30, zorder=5, 
            label='Аппроксимация', marker='x', linewidths=0.2)

plt.xlabel('x', fontsize=12)
plt.ylabel('y', fontsize=12)
plt.legend(fontsize=10)
plt.grid(True, alpha=0.3, linestyle='--')
plt.tight_layout()

plt.show()
```
